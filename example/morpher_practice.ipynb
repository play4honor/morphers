```python
import lightning.pytorch as pyt
from torch.utils.data import DataLoader
import yaml
import torch
```

```python
from src.nn import morph_data, ThingDataset, ThingLightning
```

```python
torch.set_float32_matmul_precision("medium")
```

# Data Prep

```python
cat_cols = ["blue", "dual_sim", "four_g", "three_g", "touch_screen", "wifi"]
num_cols = ["battery_power", "fc", "pc", "clock_speed", "int_memory", "m_dep",  "mobile_wt",
            "n_cores", "px_height", "px_width", "ram", "sc_h", "sc_w", "talk_time"]
target_col = "price_range"
```

```python
train_df, morphers = morph_data("data/train_new.csv", cat_cols=cat_cols, num_cols=num_cols, target_col=target_col)
```

```python
morpher_states = {feat: morpher.save_state_dict() for feat, morpher in morphers.items()}
with open("morphers.yaml", "w") as f:
    yaml.dump(morpher_states, f)
```

```python
ds = ThingDataset(train_df, morphers, target_col)
```

```python
train_ds, valid_ds = torch.utils.data.random_split(ds, [0.80, 0.20])
```

# Model

```python
train_dl = DataLoader(
    train_ds,
    batch_size=64,
    shuffle=True,
    num_workers=0,
)
```

```python
valid_dl = DataLoader(
    valid_ds,
    batch_size=64,
    shuffle=False,
    num_workers=0,
)
```

```python
trainer = pyt.Trainer(
    max_epochs = 200,
    log_every_n_steps = 50,
    logger=pyt.loggers.TensorBoardLogger("."),
    callbacks=[
        pyt.callbacks.ModelCheckpoint(
            dirpath="./model",
            save_top_k=1,
            monitor="valid_loss",
            filename="best_morpher"
        )
    ]
)
```

```python
with trainer.init_module():
    net = ThingLightning(
        morphers=morphers,
        hidden_dim=128,
        out_cats=train_df[target_col].n_unique(),
        target_col=target_col,
        optim_lr=.0002,
    )
```

```python
trainer.fit(net, train_dataloaders=train_dl, val_dataloaders=valid_dl)
```
